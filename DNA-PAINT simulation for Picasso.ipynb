```python
"""
Created on Tue Oct 20 23:43:59 2020

@author: pierr
"""

import numpy as np
import random

thickness_of_one_band = 27
distance_between_bands = 85


mu, sigma = 0, thickness_of_one_band # mean and standard deviation
#sx = 200 + np.random.normal(mu, sigma, 600)
#sy = np.random.uniform(0,1000,600)

sx = 1000 + np.random.normal(mu, sigma, 600) # 1000 in front to avoid negative numbers
sy = np.random.uniform(0,1000,600)



Struture_X = ""
Struture_Y = ""
Struture_3D = ""
Exchange_la = ""
for i in range(0,300,2):
    Struture_X = Struture_X + str(sx[i]) + "," + str(sx[i+1]+distance_between_bands) + ","
    Struture_Y = Struture_Y + str(sy[i]) + "," + str(sy[i+1]) + ","
    Struture_3D = Struture_3D + str(0) + "," + str(0) + ","
    Exchange_la = Exchange_la + str(1) + "," + str(1) + ","
    
print(Struture_X + "\n")
print(Struture_Y + "\n")
print(Struture_3D + "\n")
print(Exchange_la + "\n")
```

```python
import numpy as np

block_width = 3000          # nm
block_height = 1000        # nm
block_spacing = 170         # nm

n_blocks = 2
locs_per_block = 2000

x_all = []
y_all = []

x_start = 1000  # offset to keep coordinates positive

for i in range(n_blocks):

    x0 = x_start + i * (block_width + block_spacing)

    x = np.random.uniform(
        x0,
        x0 + block_width,
        locs_per_block
    )

    y = np.random.uniform(
        0,
        block_height,
        locs_per_block
    )

    x_all.extend(x)
    y_all.extend(y)

Structure_X = ",".join(map(str, x_all))
Structure_Y = ",".join(map(str, y_all))
Structure_3D = ",".join(["0"] * len(x_all))
Exchange_la = ",".join(["1"] * len(x_all))

print(Structure_X + "\n")
print(Structure_Y + "\n")
print(Structure_3D + "\n")
print(Exchange_la + "\n")
```
